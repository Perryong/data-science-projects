# Exploratory Data Analysis (EDA)

This notebook contains exploratory data analysis for stock market data.


```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from pathlib import Path
import sys

# Add src to path
sys.path.append(str(Path().resolve().parent / 'src'))

from data.get_data import fetch_stock_data
from data.clean_data import remove_missing_values, validate_data
from features.build_features import build_features

# Set style
sns.set_style("whitegrid")
plt.rcParams['figure.figsize'] = (12, 6)
```

## Load Data


```python
# Load configuration or set parameters directly
ticker = "AAPL"
start_date = "2020-01-01"
end_date = "2024-01-01"

# Fetch data
data = fetch_stock_data(ticker, start_date, end_date)
print(f"Data shape: {data.shape}")
print(f"\nFirst few rows:")
print(data.head())
print(f"\nData info:")
print(data.info())
print(f"\nBasic statistics:")
print(data.describe())
```

## Data Cleaning


```python
# Check for missing values
print("Missing values:")
print(data.isnull().sum())

# Clean data
cleaned_data = remove_missing_values(data, method="forward_fill")
validate_data(cleaned_data)
```

## Visualizations


```python
# Plot closing price over time
plt.figure(figsize=(14, 6))
plt.plot(cleaned_data.index, cleaned_data['Close'], label='Close Price')
plt.title(f'{ticker} Stock Price Over Time')
plt.xlabel('Date')
plt.ylabel('Price ($)')
plt.legend()
plt.grid(True)
plt.tight_layout()
plt.show()
```

```python
# Plot volume over time
plt.figure(figsize=(14, 6))
plt.bar(cleaned_data.index, cleaned_data['Volume'], alpha=0.6)
plt.title(f'{ticker} Trading Volume Over Time')
plt.xlabel('Date')
plt.ylabel('Volume')
plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()
```

```python
# Calculate and plot returns
returns = cleaned_data['Close'].pct_change().dropna()
plt.figure(figsize=(14, 6))
plt.plot(returns.index, returns, alpha=0.7)
plt.title(f'{ticker} Daily Returns')
plt.xlabel('Date')
plt.ylabel('Returns')
plt.grid(True)
plt.tight_layout()
plt.show()

print(f"Returns statistics:")
print(returns.describe())
```

## Feature Engineering


```python
# Build features
features = build_features(cleaned_data)
print(f"Features shape: {features.shape}")
print(f"\nFeature columns:")
print(features.columns.tolist())
print(f"\nFeatures info:")
print(features.info())
```
