# Model Testing Scratchpad

Quick tests and experiments for model development.


```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import sys
from pathlib import Path

# Add src to path
sys.path.append(str(Path().resolve().parent / 'src'))

from data.get_data import fetch_stock_data
from data.clean_data import remove_missing_values
from features.build_features import build_features, create_target_variable
from models.train_model import prepare_data, train_model, evaluate_model
from visualization.plot_predictions import (
    plot_predictions_vs_actual,
    plot_prediction_timeseries,
    plot_feature_importance,
    plot_metrics_comparison
)
```

## Quick Model Test


```python
# Load and prepare data
ticker = "AAPL"
data = fetch_stock_data(ticker, "2020-01-01", "2024-01-01")
cleaned_data = remove_missing_values(data, method="forward_fill")
features = build_features(cleaned_data)
target = create_target_variable(features, target_type="next_day_close")

# Prepare for training (now returns feature_names and test_dates)
X_train, X_test, y_train, y_test, scaler, feature_names, test_dates = prepare_data(features, target)

# Train model
model = train_model(X_train, y_train, model_type="random_forest", n_estimators=50)

# Evaluate (now returns predictions)
metrics, y_pred = evaluate_model(model, X_test, y_test, return_predictions=True)
print(f"\nModel Performance:")
for metric, value in metrics.items():
    print(f"{metric.upper()}: {value:.4f}")
```

## Visualizations


```python
# Comprehensive prediction visualization
plot_predictions_vs_actual(y_test, y_pred, ticker)
```

```python
# Time series plot with dates
plot_prediction_timeseries(test_dates, y_test, y_pred, ticker)
```

```python
# Feature importance (for tree-based models)
plot_feature_importance(model, feature_names, ticker, top_n=15)
```

```python
# Metrics comparison
plot_metrics_comparison(metrics, ticker)
```
